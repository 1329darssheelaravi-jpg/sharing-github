```python
# ============================================================
# VS-04 MEFSAT Batch 03 — CELL 5
# Validate canonical SMILES with RDKit
# ============================================================

!pip install rdkit
from rdkit import Chem

print("=" * 70)
print("VS-04 MEFSAT BATCH 03 — RDKit MOLECULE VALIDATION")
print("=" * 70)

# Convert canonical SMILES to RDKit Mol objects
molecules = [
    Chem.MolFromSmiles(smiles)
    for smiles in df["Canonical_SMILES"]
]

# Count parsing failures
invalid_indices = [
    i for i, mol in enumerate(molecules)
    if mol is None
]

print(f"Total compounds checked : {len(molecules)}")
print(f"Valid RDKit molecules    : {len(molecules) - len(invalid_indices)}")
print(f"Invalid RDKit molecules  : {len(invalid_indices)}")

if invalid_indices:
    print("\nInvalid molecule indices:")
    print(invalid_indices)

    print("\nCorresponding compound information:")
    display(
        df.iloc[invalid_indices][
            [
                "Serial_Number",
                "Secondary_Metabolite_Identifier",
                "Secondary_Metabolite_Name",
                "Canonical_SMILES"
            ]
        ]
    )
else:
    print("\n✓ All 275 canonical SMILES successfully parsed by RDKit.")

print("\nNo Morgan fingerprints generated.")
print("No models loaded.")
print("No predictions performed.")

print("=" * 70)
```

```python
# ============================================================
# VS-04 MEFSAT Batch 03 — CELL 6
# Generate Morgan fingerprints
# ============================================================

from rdkit.Chem import rdFingerprintGenerator
import numpy as np

print("=" * 70)
print("VS-04 MEFSAT BATCH 03 — MORGAN FINGERPRINT GENERATION")
print("=" * 70)

# Frozen fingerprint specification
MORGAN_RADIUS = 2
MORGAN_BITS = 2048

# Create Morgan fingerprint generator
morgan_generator = rdFingerprintGenerator.GetMorganGenerator(
    radius=MORGAN_RADIUS,
    fpSize=MORGAN_BITS
)

# Generate fingerprints from the already validated RDKit molecules
fingerprints = np.array([
    morgan_generator.GetFingerprintAsNumPy(mol)
    for mol in molecules
], dtype=np.uint8)

print(f"Number of compounds : {fingerprints.shape[0]}")
print(f"Number of features  : {fingerprints.shape[1]}")
print(f"Data type           : {fingerprints.dtype}")

print("\nExpected:")
print("Compounds = 275")
print("Features  = 2048")

print("\nFingerprint generation complete.")
print("No models loaded.")
print("No predictions performed.")

print("=" * 70)
```

```python
# ============================================================
# VS-04 MEFSAT Batch 03 — CELL 7
# Morgan fingerprint integrity QC
# ============================================================

print("=" * 70)
print("VS-04 MEFSAT BATCH 03 — FINGERPRINT INTEGRITY QC")
print("=" * 70)

# Basic shape checks
assert fingerprints.shape == (275, 2048), (
    f"Unexpected fingerprint shape: {fingerprints.shape}"
)

# Check for missing/non-finite values
nan_count = np.isnan(fingerprints).sum()
inf_count = np.isinf(fingerprints).sum()

# Check unique values present
unique_values = np.unique(fingerprints)

# Number of active bits per compound
active_bits = fingerprints.sum(axis=1)

print(f"Fingerprint matrix shape : {fingerprints.shape}")
print(f"NaN values               : {nan_count}")
print(f"Infinite values          : {inf_count}")
print(f"Unique values            : {unique_values}")
print(f"Minimum active bits      : {active_bits.min()}")
print(f"Maximum active bits      : {active_bits.max()}")
print(f"Mean active bits         : {active_bits.mean():.2f}")

print("\nIntegrity checks:")

if fingerprints.shape == (275, 2048):
    print("✓ Correct matrix dimensions")
else:
    print("✗ Incorrect matrix dimensions")

if nan_count == 0:
    print("✓ No NaN values")
else:
    print("✗ NaN values detected")

if inf_count == 0:
    print("✓ No infinite values")
else:
    print("✗ Infinite values detected")

if np.all(np.isin(unique_values, [0, 1])):
    print("✓ Binary fingerprint values confirmed")
else:
    print("✗ Unexpected fingerprint values detected")

if np.all(active_bits > 0):
    print("✓ Every compound has at least one active Morgan bit")
else:
    print("✗ At least one compound has zero active bits")

print("\nNo models loaded.")
print("No predictions performed.")

print("=" * 70)
```

```python
# ============================================================
# VS-04 MEFSAT Batch 03 — CELL 8
# Locate frozen VS-04 prediction models
# ============================================================

import os

BASE_DIR = "/content/drive/MyDrive/PhD_ML_Project"

print("=" * 70)
print("VS-04 MEFSAT BATCH 03 — FROZEN MODEL SEARCH")
print("=" * 70)

# Search for model files associated with VS-04
model_extensions = (".pkl", ".joblib")

model_files = []

for root, dirs, files in os.walk(BASE_DIR):
    for file in files:
        if file.lower().endswith(model_extensions):
            model_files.append(os.path.join(root, file))

print(f"\nTotal model files found: {len(model_files)}")

print("\nModel files:")
for path in sorted(model_files):
    print(path)

print("=" * 70)
```

```python
# ============================================================
# VS-04 MEFSAT Batch 03 — CELL 9
# Frozen model compatibility QC
# ============================================================

import joblib
import os

MODEL_DIR = "/content/drive/MyDrive/PhD_ML_Project/Models/XGBoost"

TARGETS = [
    "AChE",
    "BChE",
    "M2",
    "NMDAR",
    "5HT6",
    "GSK3B",
    "MAOB",
    "PDE",
    "TNFA",
    "iNOS"
]

print("=" * 70)
print("VS-04 MEFSAT BATCH 03 — FROZEN MODEL COMPATIBILITY QC")
print("=" * 70)

frozen_models = {}

for target in TARGETS:

    model_path = os.path.join(
        MODEL_DIR,
        target,
        f"{target}_XGBoost.pkl"
    )

    print(f"\n{'-' * 70}")
    print(f"Target: {target}")
    print(f"Model : {model_path}")

    if not os.path.exists(model_path):
        print("✗ MODEL FILE NOT FOUND")
        continue

    model = joblib.load(model_path)
    frozen_models[target] = model

    print(f"Model class       : {type(model).__name__}")

    # Feature-count compatibility
    if hasattr(model, "n_features_in_"):
        print(f"Expected features : {model.n_features_in_}")
    else:
        print("Expected features : attribute not available")

    # XGBoost-specific information where available
    if hasattr(model, "get_booster"):
        booster = model.get_booster()
        print(f"Booster features  : {booster.num_features()}")

    print("✓ Model loaded")
    print("✓ No prediction performed")

print("\n" + "=" * 70)
print("MODEL COMPATIBILITY SUMMARY")
print("=" * 70)

print(f"Models successfully loaded: {len(frozen_models)}")
print(f"Expected fingerprint features: {fingerprints.shape[1]}")

print("=" * 70)
```

```python
# ============================================================
# VS-04 MEFSAT Batch 03 — CELL 10
# Frozen model prediction-interface QC
# ============================================================

print("=" * 70)
print("VS-04 MEFSAT BATCH 03 — MODEL INTERFACE QC")
print("=" * 70)

for target, model in frozen_models.items():

    print(f"\n{'-' * 70}")
    print(f"Target: {target}")

    # Model classes
    if hasattr(model, "classes_"):
        print(f"Classes           : {model.classes_}")
        print(f"Number of classes : {len(model.classes_)}")
    else:
        print("Classes           : attribute not available")

    # Confirm classifier interface exists
    print(
        f"predict()         : "
        f"{'available' if hasattr(model, 'predict') else 'NOT available'}"
    )

    print(
        f"predict_proba()   : "
        f"{'available' if hasattr(model, 'predict_proba') else 'NOT available'}"
    )

print("\n" + "=" * 70)
print("NO PREDICTIONS WERE PERFORMED")
print("=" * 70)
```

```python
# ============================================================
# VS-04 MEFSAT Batch 03 — CELL 11
# Controlled AChE prediction
# ============================================================

import numpy as np

print("=" * 70)
print("VS-04 MEFSAT BATCH 03 — AChE PREDICTION")
print("=" * 70)

# Retrieve the already-loaded frozen AChE model
ache_model = frozen_models["AChE"]

# ------------------------------------------------------------
# Perform prediction
# ------------------------------------------------------------

ache_pred = ache_model.predict(fingerprints)
ache_proba = ache_model.predict_proba(fingerprints)

# Extract probability for class 1
ache_prob_class1 = ache_proba[:, 1]

# ------------------------------------------------------------
# Basic output validation
# ------------------------------------------------------------

print(f"Number of compounds predicted : {len(ache_pred)}")
print(f"Prediction array shape         : {ache_pred.shape}")
print(f"Probability array shape        : {ache_proba.shape}")

print("\nPredicted classes:")
print(np.unique(ache_pred, return_counts=True))

print("\nClass 1 probability range:")
print(f"Minimum : {ache_prob_class1.min():.6f}")
print(f"Maximum : {ache_prob_class1.max():.6f}")
print(f"Mean    : {ache_prob_class1.mean():.6f}")

# ------------------------------------------------------------
# Create QC table — raw predictions only
# ------------------------------------------------------------

ache_qc = df[
    [
        "Serial_Number",
        "Secondary_Metabolite_Identifier",
        "Secondary_Metabolite_Name"
    ]
].copy()

ache_qc["AChE_Prediction"] = ache_pred
ache_qc["AChE_Probability_Class1"] = ache_prob_class1

print("\nFirst 10 AChE predictions:")
display(ache_qc.head(10))

print("=" * 70)
print("AChE prediction completed.")
print("No hit threshold applied.")
print("No compounds excluded.")
print("No ranking performed.")
print("=" * 70)
```

```python
# ============================================================
# VS-04 MEFSAT Batch 03 — CELL 12
# AChE probability distribution QC
# ============================================================

print("=" * 70)
print("VS-04 MEFSAT BATCH 03 — AChE PROBABILITY QC")
print("=" * 70)

# Sort raw AChE probabilities from highest to lowest
ache_probability_qc = ache_qc.sort_values(
    "AChE_Probability_Class1",
    ascending=False
).reset_index(drop=True)

# Display the top 20 compounds by raw class-1 probability
print("\nTop 20 compounds by AChE class-1 probability:")
display(
    ache_probability_qc[
        [
            "Serial_Number",
            "Secondary_Metabolite_Identifier",
            "Secondary_Metabolite_Name",
            "AChE_Prediction",
            "AChE_Probability_Class1"
        ]
    ].head(20)
)

# Probability summary
print("\nProbability summary:")
print(
    ache_qc["AChE_Probability_Class1"].describe(
        percentiles=[0.50, 0.75, 0.90, 0.95, 0.99]
    )
)

print("\nMaximum probability:")
max_idx = ache_qc["AChE_Probability_Class1"].idxmax()
print(
    ache_qc.loc[
        max_idx,
        [
            "Serial_Number",
            "Secondary_Metabolite_Identifier",
            "Secondary_Metabolite_Name",
            "AChE_Probability_Class1"
        ]
    ]
)

print("=" * 70)
```

```python
# ============================================================
# VS-04 MEFSAT Batch 03 — CELL 13
# Create AChE raw prediction table
# ============================================================

ache_results = df.copy()

ache_results["AChE_Prediction"] = ache_pred
ache_results["AChE_Probability_Class1"] = ache_prob_class1

print("=" * 70)
print("VS-04 MEFSAT BATCH 03 — AChE RAW RESULTS")
print("=" * 70)

print(f"Rows                    : {len(ache_results)}")
print(f"AChE prediction column  : {'AChE_Prediction' in ache_results.columns}")
print(f"AChE probability column : {'AChE_Probability_Class1' in ache_results.columns}")

print("\nPrediction counts:")
print(ache_results["AChE_Prediction"].value_counts().sort_index())

print("\nMissing AChE predictions:")
print(ache_results["AChE_Prediction"].isna().sum())

print("\nMissing AChE probabilities:")
print(ache_results["AChE_Probability_Class1"].isna().sum())

print("\nAChE probability range:")
print(
    f"{ache_results['AChE_Probability_Class1'].min():.6f}"
    " to "
    f"{ache_results['AChE_Probability_Class1'].max():.6f}"
)

print("\nFirst 5 rows:")
display(
    ache_results[
        [
            "Secondary_Metabolite_Identifier",
            "Secondary_Metabolite_Name",
            "AChE_Prediction",
            "AChE_Probability_Class1"
        ]
    ].head()
)

print("=" * 70)
print("AChE raw result table created.")
print("No additional threshold applied.")
print("=" * 70)
```

```python
# ============================================================
# VS-04 MEFSAT Batch 03 — CELL 14
# One-shot prediction for remaining 9 frozen models
# ============================================================

print("=" * 70)
print("VS-04 MEFSAT BATCH 03 — MULTI-TARGET PREDICTION")
print("=" * 70)

# Remaining frozen targets
REMAINING_TARGETS = [
    "BChE",
    "M2",
    "NMDAR",
    "5HT6",
    "GSK3B",
    "MAOB",
    "PDE",
    "TNFA",
    "iNOS"
]

# Dictionary to store raw predictions
prediction_results = {}

for target in REMAINING_TARGETS:

    print(f"\n{'-' * 70}")
    print(f"Predicting: {target}")

    model = frozen_models[target]

    # Frozen model prediction
    predictions = model.predict(fingerprints)
    probabilities = model.predict_proba(fingerprints)

    # Class-1 probability
    probability_class1 = probabilities[:, 1]

    # Store raw outputs
    prediction_results[target] = {
        "prediction": predictions,
        "probability_class1": probability_class1
    }

    # Immediate integrity checks
    assert len(predictions) == 275
    assert probabilities.shape == (275, 2)
    assert len(probability_class1) == 275
    assert np.all(np.isfinite(probability_class1))

    print(f"Compounds predicted : {len(predictions)}")
    print(f"Probability shape   : {probabilities.shape}")

    unique_classes, class_counts = np.unique(
        predictions,
        return_counts=True
    )

    print("Prediction counts   :")
    for cls, count in zip(unique_classes, class_counts):
        print(f"  Class {cls}: {count}")

    print(
        f"Class-1 probability : "
        f"{probability_class1.min():.6f} - "
        f"{probability_class1.max():.6f}"
    )

print("\n" + "=" * 70)
print("MULTI-TARGET PREDICTION SUMMARY")
print("=" * 70)

print(f"Targets processed : {len(prediction_results)}")

for target in REMAINING_TARGETS:
    pred = prediction_results[target]["prediction"]
    prob = prediction_results[target]["probability_class1"]

    print(
        f"{target:7s} | "
        f"n={len(pred):3d} | "
        f"class-1={np.sum(pred == 1):3d} | "
        f"class-0={np.sum(pred == 0):3d} | "
        f"max P(1)={prob.max():.6f}"
    )

print("=" * 70)
print("All 9 remaining frozen models predicted.")
print("No hit threshold applied.")
print("No compounds excluded.")
print("No ranking performed.")
print("=" * 70)
```

```python
# ============================================================
# VS-04 MEFSAT Batch 03 — CELL 15
# Build master raw prediction table
# ============================================================

print("=" * 70)
print("VS-04 MEFSAT BATCH 03 — MASTER RAW PREDICTION TABLE")
print("=" * 70)

# Start from the original validated 275-compound library
master_predictions = df.copy()

# ------------------------------------------------------------
# Add AChE results
# ------------------------------------------------------------

master_predictions["AChE_Prediction"] = ache_pred
master_predictions["AChE_Probability_Class1"] = ache_prob_class1

# ------------------------------------------------------------
# Add the remaining 9 target results
# ------------------------------------------------------------

for target in REMAINING_TARGETS:

    master_predictions[f"{target}_Prediction"] = (
        prediction_results[target]["prediction"]
    )

    master_predictions[f"{target}_Probability_Class1"] = (
        prediction_results[target]["probability_class1"]
    )

# ------------------------------------------------------------
# Master-table integrity checks
# ------------------------------------------------------------

print(f"Rows                    : {len(master_predictions)}")

expected_prediction_columns = [
    f"{target}_Prediction"
    for target in TARGETS
]

expected_probability_columns = [
    f"{target}_Probability_Class1"
    for target in TARGETS
]

missing_prediction_columns = [
    col for col in expected_prediction_columns
    if col not in master_predictions.columns
]

missing_probability_columns = [
    col for col in expected_probability_columns
    if col not in master_predictions.columns
]

print(f"Prediction columns      : {len(expected_prediction_columns)}")
print(f"Probability columns     : {len(expected_probability_columns)}")

print("\nMissing prediction columns:")
print(missing_prediction_columns)

print("\nMissing probability columns:")
print(missing_probability_columns)

# Check every target has 275 predictions and 275 probabilities
print("\nTarget-level row checks:")

for target in TARGETS:

    pred_col = f"{target}_Prediction"
    prob_col = f"{target}_Probability_Class1"

    assert master_predictions[pred_col].notna().all()
    assert master_predictions[prob_col].notna().all()
    assert len(master_predictions[pred_col]) == 275
    assert len(master_predictions[prob_col]) == 275

    print(
        f"✓ {target:7s} | "
        f"predictions={master_predictions[pred_col].notna().sum():3d} | "
        f"probabilities={master_predictions[prob_col].notna().sum():3d}"
    )

print("\nFirst 5 rows of prediction columns:")

display(
    master_predictions[
        ["Secondary_Metabolite_Identifier"] +
        expected_prediction_columns +
        expected_probability_columns
    ].head()
)

print("=" * 70)
print("MASTER RAW PREDICTION TABLE CREATED")
print("No hit threshold applied.")
print("No compounds excluded.")
print("No ranking performed.")
print("=" * 70)
```

```python
# ============================================================
# VS-04 MEFSAT Batch 03 — CELL 16
# Inspect existing screening-decision fields
# ============================================================

print("=" * 70)
print("VS-04 MEFSAT BATCH 03 — SCREENING DECISION QC")
print("=" * 70)

print("\nExisting Screening_Decision values:")
print(
    master_predictions["Screening_Decision"]
    .value_counts(dropna=False)
)

print("\nExisting Structural_QC values:")
print(
    master_predictions["Structural_QC"]
    .value_counts(dropna=False)
)

print("\nExisting Exclusion_Reason values:")
print(
    master_predictions["Exclusion_Reason"]
    .value_counts(dropna=False)
)

print("\nDecision-related columns:")
decision_columns = [
    col for col in master_predictions.columns
    if any(term in col.lower() for term in [
        "decision",
        "exclude",
        "hold",
        "screen",
        "qc"
    ])
]

for col in decision_columns:
    print(f" - {col}")

print("=" * 70)
```

```python
# ============================================================
# VS-04 MEFSAT Batch 03 — CELL 17
# Extract predicted receptor hits
# ============================================================

print("=" * 70)
print("VS-04 MEFSAT BATCH 03 — PREDICTED HIT EXTRACTION")
print("=" * 70)

# Receptor-level prediction columns
prediction_columns = {
    "AChE": "AChE_Prediction",
    "BChE": "BChE_Prediction",
    "M2": "M2_Prediction",
    "NMDAR": "NMDAR_Prediction",
    "5HT6": "5HT6_Prediction",
    "GSK3B": "GSK3B_Prediction",
    "MAOB": "MAOB_Prediction",
    "PDE": "PDE_Prediction",
    "TNFA": "TNFA_Prediction",
    "iNOS": "iNOS_Prediction"
}

probability_columns = {
    "AChE": "AChE_Probability_Class1",
    "BChE": "BChE_Probability_Class1",
    "M2": "M2_Probability_Class1",
    "NMDAR": "NMDAR_Probability_Class1",
    "5HT6": "5HT6_Probability_Class1",
    "GSK3B": "GSK3B_Probability_Class1",
    "MAOB": "MAOB_Probability_Class1",
    "PDE": "PDE_Probability_Class1",
    "TNFA": "TNFA_Probability_Class1",
    "iNOS": "iNOS_Probability_Class1"
}

hit_tables = []

for target in TARGETS:

    pred_col = prediction_columns[target]
    prob_col = probability_columns[target]

    # Select compounds predicted as class 1
    target_hits = master_predictions[
        master_predictions[pred_col] == 1
    ].copy()

    # Keep relevant information
    target_hits = target_hits[
        [
            "Serial_Number",
            "Secondary_Metabolite_Identifier",
            "Secondary_Metabolite_Name",
            "SMILES",
            "Canonical_SMILES",
            pred_col,
            prob_col
        ]
    ].copy()

    # Standardise output column names
    target_hits = target_hits.rename(
        columns={
            pred_col: "Predicted_Class",
            prob_col: "Active_Probability"
        }
    )

    # Add receptor
    target_hits.insert(
        5,
        "Target",
        target
    )

    hit_tables.append(target_hits)

# Combine all receptor-level hits
if hit_tables:
    hit_table = pd.concat(
        hit_tables,
        ignore_index=True
    )
else:
    hit_table = pd.DataFrame()

print("\nPredicted hits by target:")
print(
    master_predictions[
        list(prediction_columns.values())
    ].sum()
)

print("\nTotal predicted hit rows:")
print(len(hit_table))

print("\nHit table:")
if len(hit_table) > 0:
    display(hit_table)
else:
    print("No predicted hits identified.")

print("=" * 70)
print("No probability threshold applied.")
print("Hit criterion: Predicted_Class == 1")
print("=" * 70)
```

```python
# ============================================================
# VS-04 — CELL 21A
# Validate Table 2 ↔ MEFSAT exact compound-name matches
# ============================================================

print("=" * 70)
print("VS-04 — TABLE 2 ↔ MEFSAT EXACT MATCH VALIDATION")
print("=" * 70)

# ------------------------------------------------------------
# Create normalised names
# ------------------------------------------------------------

table2_names = (
    table2["Verified_Name"]
    .astype(str)
    .str.strip()
    .str.lower()
)

mefsat_names = (
    mefsat["Secondary_Metabolite_Name"]
    .astype(str)
    .str.strip()
    .str.lower()
)

# ------------------------------------------------------------
# Find exact overlaps
# ------------------------------------------------------------

overlap_names = sorted(
    set(table2_names) & set(mefsat_names)
)

print(f"\nExact name matches: {len(overlap_names)}")

# ------------------------------------------------------------
# Build mapping table WITHOUT merge()
# ------------------------------------------------------------

matched_rows = []

for name in overlap_names:

    table2_match = table2.loc[
        table2_names == name
    ].iloc[0]

    mefsat_match = mefsat.loc[
        mefsat_names == name
    ].iloc[0]

    matched_rows.append({
        "Table2_ID": table2_match["Table2_ID"],
        "Table2_Name": table2_match["Verified_Name"],
        "MEFSAT_MSID": mefsat_match[
            "Secondary_Metabolite_Identifier"
        ],
        "MEFSAT_Name": mefsat_match[
            "Secondary_Metabolite_Name"
        ]
    })

overlap_table = pd.DataFrame(matched_rows)

# ------------------------------------------------------------
# Display mapping
# ------------------------------------------------------------

print("\nMatched compounds:")
display(overlap_table)

# ------------------------------------------------------------
# Integrity checks
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("MATCH INTEGRITY CHECKS")
print("=" * 70)

print(
    "Unique Table 2 IDs matched :",
    overlap_table["Table2_ID"].nunique()
)

print(
    "Unique MEFSAT MSIDs matched:",
    overlap_table["MEFSAT_MSID"].nunique()
)

print(
    "Duplicate Table 2 IDs      :",
    overlap_table["Table2_ID"].duplicated().sum()
)

print(
    "Duplicate MEFSAT MSIDs     :",
    overlap_table["MEFSAT_MSID"].duplicated().sum()
)

print("=" * 70)
print("NO DATA MERGED")
print("NO ORIGINAL DATA MODIFIED")
print("=" * 70)
```

```python
# ============================================================
# VS-04 — CELL 22
# Compare prediction classes for 15 overlapping compounds
# ============================================================

print("=" * 70)
print("VS-04 — OVERLAPPING COMPOUNDS: PREDICTION CLASS COMPARISON")
print("=" * 70)

# ------------------------------------------------------------
# Define the 10 targets
# ------------------------------------------------------------

TARGETS = [
    "AChE",
    "BChE",
    "M2",
    "NMDAR",
    "5HT6",
    "GSK3B",
    "MAOB",
    "PDE",
    "TNFA",
    "iNOS"
]

# ------------------------------------------------------------
# Identify Table 2 prediction columns
# ------------------------------------------------------------

table2_prediction_columns = {
    target: f"{target}_Prediction"
    for target in TARGETS
}

# Verify columns exist
missing_table2 = [
    col for col in table2_prediction_columns.values()
    if col not in table2.columns
]

print("\nMissing Table 2 prediction columns:")
print(missing_table2)

# ------------------------------------------------------------
# Create comparison table using validated mapping
# ------------------------------------------------------------

comparison_class = overlap_table[
    [
        "Table2_ID",
        "Table2_Name",
        "MEFSAT_MSID",
        "MEFSAT_Name"
    ]
].copy()

# ------------------------------------------------------------
# Add prediction classes from both datasets
# ------------------------------------------------------------

for target in TARGETS:

    table2_id = comparison_class["Table2_ID"]

    # Table 2 predictions
    table2_lookup = (
        table2[
            ["Table2_ID", table2_prediction_columns[target]]
        ]
        .drop_duplicates("Table2_ID")
        .set_index("Table2_ID")
    )

    comparison_class[f"{target}_Table2"] = (
        table2_id.map(
            table2_lookup[
                table2_prediction_columns[target]
            ]
        )
    )

    # MEFSAT predictions
    mefsat_lookup = (
        mefsat[
            [
                "Secondary_Metabolite_Identifier",
                f"{target}_Prediction"
            ]
        ]
        .drop_duplicates(
            "Secondary_Metabolite_Identifier"
        )
        .set_index(
            "Secondary_Metabolite_Identifier"
        )
    )

    comparison_class[f"{target}_MEFSAT"] = (
        comparison_class["MEFSAT_MSID"].map(
            mefsat_lookup[f"{target}_Prediction"]
        )
    )

# ------------------------------------------------------------
# QC: no missing predictions
# ------------------------------------------------------------

prediction_columns = [
    col for col in comparison_class.columns
    if col.endswith("_Table2") or col.endswith("_MEFSAT")
]

print("\nMissing prediction values:")
print(
    comparison_class[prediction_columns].isna().sum()
)

# ------------------------------------------------------------
# Calculate agreement for each compound
# ------------------------------------------------------------

agreement_counts = []

for _, row in comparison_class.iterrows():

    agreements = 0

    for target in TARGETS:

        t2 = row[f"{target}_Table2"]
        mf = row[f"{target}_MEFSAT"]

        if t2 == mf:
            agreements += 1

    agreement_counts.append(agreements)

comparison_class["Number_of_Targets_Agreeing"] = (
    agreement_counts
)

comparison_class["Number_of_Targets_Differing"] = (
    len(TARGETS) -
    comparison_class["Number_of_Targets_Agreeing"]
)

# ------------------------------------------------------------
# Display
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("PREDICTION CLASS COMPARISON")
print("=" * 70)

display(comparison_class)

# ------------------------------------------------------------
# Overall agreement summary
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("AGREEMENT SUMMARY")
print("=" * 70)

print(
    f"Overlapping compounds : {len(comparison_class)}"
)

print(
    f"Targets compared      : {len(TARGETS)}"
)

print(
    f"Total comparisons     : "
    f"{len(comparison_class) * len(TARGETS)}"
)

total_agreements = (
    comparison_class["Number_of_Targets_Agreeing"]
    .sum()
)

total_differences = (
    comparison_class["Number_of_Targets_Differing"]
    .sum()
)

print(f"Agreements             : {total_agreements}")
print(f"Differences            : {total_differences}")

print("\nAgreement per compound:")
display(
    comparison_class[
        [
            "Table2_ID",
            "Table2_Name",
            "MEFSAT_MSID",
            "Number_of_Targets_Agreeing",
            "Number_of_Targets_Differing"
        ]
    ]
)

print("=" * 70)
print("NO SOURCE DATA MODIFIED")
print("=" * 70)
```

```python
# ============================================================
# VS-04 — CELL 23
# Compare active probabilities for 15 overlapping compounds
# ============================================================

print("=" * 70)
print("VS-04 — OVERLAPPING COMPOUNDS: PROBABILITY COMPARISON")
print("=" * 70)

# ------------------------------------------------------------
# Identify probability columns
# ------------------------------------------------------------

table2_probability_columns = {
    target: f"{target}_Probability"
    for target in TARGETS
}

mefsat_probability_columns = {
    target: f"{target}_Probability_Class1"
    for target in TARGETS
}

# Check Table 2 columns
missing_table2_probability = [
    col
    for col in table2_probability_columns.values()
    if col not in table2.columns
]

# Check MEFSAT columns
missing_mefsat_probability = [
    col
    for col in mefsat_probability_columns.values()
    if col not in mefsat.columns
]

print("\nMissing Table 2 probability columns:")
print(missing_table2_probability)

print("\nMissing MEFSAT probability columns:")
print(missing_mefsat_probability)

# ------------------------------------------------------------
# Build probability comparison table
# ------------------------------------------------------------

comparison_probability = overlap_table[
    [
        "Table2_ID",
        "Table2_Name",
        "MEFSAT_MSID",
        "MEFSAT_Name"
    ]
].copy()

# ------------------------------------------------------------
# Add probabilities
# ------------------------------------------------------------

for target in TARGETS:

    # Table 2 lookup
    table2_lookup = (
        table2[
            [
                "Table2_ID",
                table2_probability_columns[target]
            ]
        ]
        .drop_duplicates("Table2_ID")
        .set_index("Table2_ID")
    )

    comparison_probability[
        f"{target}_Table2"
    ] = comparison_probability["Table2_ID"].map(
        table2_lookup[
            table2_probability_columns[target]
        ]
    )

    # MEFSAT lookup
    mefsat_lookup = (
        mefsat[
            [
                "Secondary_Metabolite_Identifier",
                mefsat_probability_columns[target]
            ]
        ]
        .drop_duplicates(
            "Secondary_Metabolite_Identifier"
        )
        .set_index(
            "Secondary_Metabolite_Identifier"
        )
    )

    comparison_probability[
        f"{target}_MEFSAT"
    ] = comparison_probability["MEFSAT_MSID"].map(
        mefsat_lookup[
            mefsat_probability_columns[target]
        ]
    )

# ------------------------------------------------------------
# Missing-value QC
# ------------------------------------------------------------

probability_columns = [
    col
    for col in comparison_probability.columns
    if col.endswith("_Table2")
    or col.endswith("_MEFSAT")
]

print("\nMissing probability values:")
print(
    comparison_probability[
        probability_columns
    ].isna().sum()
)

# ------------------------------------------------------------
# Calculate absolute differences
# ------------------------------------------------------------

difference_columns = []

for target in TARGETS:

    table2_col = f"{target}_Table2"
    mefsat_col = f"{target}_MEFSAT"
    difference_col = f"{target}_Absolute_Difference"

    comparison_probability[difference_col] = (
        comparison_probability[table2_col]
        -
        comparison_probability[mefsat_col]
    ).abs()

    difference_columns.append(difference_col)

# Mean absolute difference per compound
comparison_probability["Mean_Absolute_Difference"] = (
    comparison_probability[difference_columns].mean(axis=1)
)

# Maximum difference per compound
comparison_probability["Maximum_Absolute_Difference"] = (
    comparison_probability[difference_columns].max(axis=1)
)

# ------------------------------------------------------------
# Display probability comparison
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("PROBABILITY COMPARISON")
print("=" * 70)

display(comparison_probability)

# ------------------------------------------------------------
# Overall probability agreement summary
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("PROBABILITY AGREEMENT SUMMARY")
print("=" * 70)

all_differences = (
    comparison_probability[difference_columns]
    .to_numpy()
)

print(
    f"Total probability comparisons : "
    f"{all_differences.size}"
)

print(
    f"Mean absolute difference      : "
    f"{all_differences.mean():.6f}"
)

print(
    f"Maximum absolute difference   : "
    f"{all_differences.max():.6f}"
)

print(
    f"Minimum absolute difference   : "
    f"{all_differences.min():.6f}"
)

print("\nMean probability difference by compound:")

display(
    comparison_probability[
        [
            "Table2_ID",
            "Table2_Name",
            "MEFSAT_MSID",
            "Mean_Absolute_Difference",
            "Maximum_Absolute_Difference"
        ]
    ]
    .sort_values(
        "Mean_Absolute_Difference",
        ascending=False
    )
)

print("=" * 70)
print("NO SOURCE DATA MODIFIED")
print("=" * 70)
```

```python
# ============================================================
# VS-04 — CELL 25
# Extract all predicted target associations
# ============================================================

print("=" * 70)
print("VS-04 — COMBINED GANODERMA PREDICTED-HIT EXTRACTION")
print("=" * 70)

# ------------------------------------------------------------
# Extract predicted hits
# ------------------------------------------------------------

hit_rows = []

for _, row in combined_predictions.iterrows():

    for target in TARGETS:

        prediction_col = f"{target}_Prediction"
        probability_col = f"{target}_Probability_Class1"

        if row[prediction_col] == 1:

            hit_rows.append({
                "Combined_ID": row["Combined_ID"],
                "Compound_Name": row["Combined_Name"],
                "Prediction_Source": row["Prediction_Source"],
                "Target": target,
                "Predicted_Class": int(row[prediction_col]),
                "Active_Probability": row[probability_col]
            })

combined_hits = pd.DataFrame(hit_rows)

# ------------------------------------------------------------
# Handle empty result safely
# ------------------------------------------------------------

if combined_hits.empty:

    print("\nNo predicted class-1 associations found.")

else:

    # --------------------------------------------------------
    # Sort by compound, then target
    # --------------------------------------------------------

    combined_hits = combined_hits.sort_values(
        [
            "Combined_ID",
            "Target"
        ]
    ).reset_index(drop=True)

    # --------------------------------------------------------
    # Summary
    # --------------------------------------------------------

    print("\nPredicted hits by target:")

    target_hit_counts = (
        combined_hits["Target"]
        .value_counts()
        .reindex(TARGETS, fill_value=0)
    )

    print(target_hit_counts)

    print(
        f"\nTotal predicted compound-target associations : "
        f"{len(combined_hits)}"
    )

    print(
        f"Unique compounds among hits                 : "
        f"{combined_hits['Combined_ID'].nunique()}"
    )

    print(
        f"Unique targets among hits                   : "
        f"{combined_hits['Target'].nunique()}"
    )

    # --------------------------------------------------------
    # Source distribution
    # --------------------------------------------------------

    print("\nPredicted hits by source:")

    print(
        combined_hits[
            "Prediction_Source"
        ].value_counts()
    )

    # --------------------------------------------------------
    # Display all hits
    # --------------------------------------------------------

    print("\n" + "=" * 70)
    print("ALL COMBINED PREDICTED HITS")
    print("=" * 70)

    display(combined_hits)

    # --------------------------------------------------------
    # Probability summary
    # --------------------------------------------------------

    print("\n" + "=" * 70)
    print("ACTIVE-PROBABILITY SUMMARY")
    print("=" * 70)

    print(
        combined_hits[
            "Active_Probability"
        ].describe()
    )

# ------------------------------------------------------------
# Final QC
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("PREDICTED-HIT EXTRACTION COMPLETE")
print("No probability threshold applied.")
print("Hit criterion: Predicted_Class == 1")
print("No compounds excluded.")
print("No ranking performed.")
print("=" * 70)
```

```python
# ============================================================
# VS-04 — CELL 27
# Full Ganoderma compound × receptor prediction matrix
# ============================================================

print("=" * 70)
print("VS-04 — FULL GANODERMA × RECEPTOR PREDICTION MATRIX")
print("=" * 70)

# ------------------------------------------------------------
# Build activity matrix
# ------------------------------------------------------------

activity_matrix = combined_predictions[
    [
        "Combined_ID",
        "Combined_Name",
        "Prediction_Source"
    ]
].copy()

# Add prediction class for each target
for target in TARGETS:
    activity_matrix[target] = combined_predictions[
        f"{target}_Prediction"
    ].astype(int)

# ------------------------------------------------------------
# Add number of predicted targets per compound
# ------------------------------------------------------------

activity_matrix["Number_of_Predicted_Targets"] = (
    activity_matrix[TARGETS].sum(axis=1)
)

# ------------------------------------------------------------
# QC
# ------------------------------------------------------------

print("\nMatrix dimensions:")
print(f"Rows    : {len(activity_matrix)}")
print(f"Columns : {len(activity_matrix.columns)}")

print("\nExpected rows:")
print("290")

print("\nDuplicate compound IDs:")
print(
    activity_matrix["Combined_ID"].duplicated().sum()
)

print("\nMissing values:")
print(
    activity_matrix[
        TARGETS
    ].isna().sum()
)

# ------------------------------------------------------------
# Binary-value validation
# ------------------------------------------------------------

invalid_values = {}

for target in TARGETS:

    values = set(
        activity_matrix[target].dropna().unique()
    )

    invalid = values - {0, 1}

    if invalid:
        invalid_values[target] = invalid

print("\nInvalid prediction values:")
print(invalid_values)

# ------------------------------------------------------------
# Target-level predicted activity counts
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("PREDICTED ACTIVITY COUNT BY TARGET")
print("=" * 70)

target_counts = (
    activity_matrix[TARGETS]
    .sum()
    .sort_values(ascending=False)
)

print(target_counts)

# ------------------------------------------------------------
# Compound-level predicted target counts
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("COMPOUND TARGET-COUNT DISTRIBUTION")
print("=" * 70)

print(
    activity_matrix[
        "Number_of_Predicted_Targets"
    ].value_counts()
    .sort_index()
)

# ------------------------------------------------------------
# Compounds with at least one predicted target
# ------------------------------------------------------------

predicted_compounds = activity_matrix[
    activity_matrix[
        "Number_of_Predicted_Targets"
    ] > 0
].copy()

print(
    "\nCompounds with ≥1 predicted target:",
    len(predicted_compounds)
)

print(
    "Compounds with no predicted targets:",
    len(activity_matrix) -
    len(predicted_compounds)
)

# ------------------------------------------------------------
# Display compounds with predicted activity
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("COMPOUNDS WITH PREDICTED TARGET ACTIVITY")
print("=" * 70)

display(
    predicted_compounds.sort_values(
        [
            "Number_of_Predicted_Targets",
            "Combined_Name"
        ],
        ascending=[False, True]
    )
)

# ------------------------------------------------------------
# Final QC
# ------------------------------------------------------------

matrix_qc_pass = (
    len(activity_matrix) == 290
    and activity_matrix[
        "Combined_ID"
    ].duplicated().sum() == 0
    and activity_matrix[
        TARGETS
    ].isna().sum().sum() == 0
    and len(invalid_values) == 0
)

print("\n" + "=" * 70)

if matrix_qc_pass:
    print("✓ FULL ACTIVITY MATRIX QC PASSED")
else:
    print("✗ FULL ACTIVITY MATRIX QC FAILED")

print("=" * 70)
print("No new predictions performed.")
print("No thresholds applied.")
print("No source data modified.")
print("=" * 70)
```

```python
# ============================================================
# VS-04 — CELL 28
# Full Ganoderma × receptor active-probability matrix
# ============================================================

print("=" * 70)
print("VS-04 — FULL GANODERMA × RECEPTOR ACTIVE-PROBABILITY MATRIX")
print("=" * 70)

# ------------------------------------------------------------
# Build probability matrix using the same 290 compounds
# ------------------------------------------------------------

probability_matrix = combined_predictions[
    [
        "Combined_ID",
        "Combined_Name",
        "Prediction_Source"
    ]
].copy()

# Add class-1 probability for each target
for target in TARGETS:

    probability_matrix[target] = combined_predictions[
        f"{target}_Probability_Class1"
    ].astype(float)

# ------------------------------------------------------------
# QC: dimensions
# ------------------------------------------------------------

print("\nMatrix dimensions:")
print(f"Rows    : {len(probability_matrix)}")
print(f"Columns : {len(probability_matrix.columns)}")

print("\nExpected rows:")
print("290")

# ------------------------------------------------------------
# QC: duplicate compounds
# ------------------------------------------------------------

print("\nDuplicate compound IDs:")
print(
    probability_matrix[
        "Combined_ID"
    ].duplicated().sum()
)

# ------------------------------------------------------------
# QC: missing values
# ------------------------------------------------------------

print("\nMissing probability values:")

missing_probabilities = (
    probability_matrix[TARGETS]
    .isna()
    .sum()
)

print(missing_probabilities)

# ------------------------------------------------------------
# QC: probability range
# ------------------------------------------------------------

invalid_probability_values = {}

for target in TARGETS:

    values = probability_matrix[target]

    invalid = values[
        (values < 0) |
        (values > 1)
    ]

    if len(invalid) > 0:
        invalid_probability_values[target] = len(invalid)

print("\nProbability values outside [0, 1]:")
print(invalid_probability_values)

# ------------------------------------------------------------
# Overall probability range
# ------------------------------------------------------------

overall_min = (
    probability_matrix[TARGETS]
    .min()
    .min()
)

overall_max = (
    probability_matrix[TARGETS]
    .max()
    .max()
)

print("\nOverall probability range:")
print(f"Minimum : {overall_min:.6f}")
print(f"Maximum : {overall_max:.6f}")

# ------------------------------------------------------------
# Target-level probability summary
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("TARGET-LEVEL PROBABILITY SUMMARY")
print("=" * 70)

target_probability_summary = pd.DataFrame({
    "Minimum": probability_matrix[TARGETS].min(),
    "Maximum": probability_matrix[TARGETS].max(),
    "Mean": probability_matrix[TARGETS].mean(),
    "Median": probability_matrix[TARGETS].median(),
    "95th_Percentile": probability_matrix[TARGETS].quantile(0.95)
})

display(target_probability_summary)

# ------------------------------------------------------------
# Check correspondence with activity matrix
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("ACTIVITY ↔ PROBABILITY MATRIX CORRESPONDENCE")
print("=" * 70)

same_ids = (
    probability_matrix["Combined_ID"].tolist()
    ==
    activity_matrix["Combined_ID"].tolist()
)

same_names = (
    probability_matrix["Combined_Name"].tolist()
    ==
    activity_matrix["Combined_Name"].tolist()
)

print(
    "Compound ID order identical:",
    same_ids
)

print(
    "Compound name order identical:",
    same_names
)

# ------------------------------------------------------------
# Display first 10 rows
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("FIRST 10 COMPOUNDS")
print("=" * 70)

display(
    probability_matrix.head(10)
)

# ------------------------------------------------------------
# Final QC
# ------------------------------------------------------------

probability_qc_pass = (
    len(probability_matrix) == 290
    and probability_matrix[
        "Combined_ID"
    ].duplicated().sum() == 0
    and missing_probabilities.sum() == 0
    and len(invalid_probability_values) == 0
    and same_ids
    and same_names
)

print("\n" + "=" * 70)

if probability_qc_pass:
    print("✓ FULL PROBABILITY MATRIX QC PASSED")
else:
    print("✗ FULL PROBABILITY MATRIX QC FAILED")

print("=" * 70)
print("No new predictions performed.")
print("No probability threshold applied.")
print("No source data modified.")
print("=" * 70)
```

```python
# ============================================================
# VS-04 — CELL 29
# Top 10 compounds by active probability for each target
# ============================================================

print("=" * 70)
print("VS-04 — TOP 10 COMPOUNDS BY ACTIVE PROBABILITY")
print("=" * 70)

top10_by_target = {}

for target in TARGETS:

    target_table = probability_matrix[
        [
            "Combined_ID",
            "Combined_Name",
            "Prediction_Source",
            target
        ]
    ].copy()

    target_table = target_table.sort_values(
        by=target,
        ascending=False
    ).reset_index(drop=True)

    target_table.insert(
        0,
        "Rank",
        range(1, len(target_table) + 1)
    )

    target_table = target_table.rename(
        columns={target: "Active_Probability"}
    )

    # Add predicted class from validated activity matrix
    target_table["Predicted_Class"] = (
        activity_matrix
        .set_index("Combined_ID")
        .loc[
            target_table["Combined_ID"],
            target
        ]
        .values
    )

    # Keep top 10 only
    top10_by_target[target] = target_table.head(10).copy()

    print("\n" + "-" * 70)
    print(f"TARGET: {target}")
    print("-" * 70)

    display(
        top10_by_target[target]
    )

# ------------------------------------------------------------
# Create one combined top-10 table
# ------------------------------------------------------------

top10_combined = pd.concat(
    [
        table.assign(Target=target)
        for target, table in top10_by_target.items()
    ],
    ignore_index=True
)

# Reorder columns
top10_combined = top10_combined[
    [
        "Target",
        "Rank",
        "Combined_ID",
        "Combined_Name",
        "Prediction_Source",
        "Predicted_Class",
        "Active_Probability"
    ]
]

# ------------------------------------------------------------
# QC
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("TOP-10 RANKING QC")
print("=" * 70)

print(
    "Number of targets:",
    top10_combined["Target"].nunique()
)

print(
    "Rows in combined top-10 table:",
    len(top10_combined)
)

print(
    "Expected rows:",
    len(TARGETS) * 10
)

print(
    "Missing values:"
)

print(
    top10_combined[
        [
            "Target",
            "Rank",
            "Combined_ID",
            "Combined_Name",
            "Prediction_Source",
            "Predicted_Class",
            "Active_Probability"
        ]
    ].isna().sum()
)

# Check rank integrity
rank_check = (
    top10_combined
    .groupby("Target")["Rank"]
    .apply(lambda x: list(x) == list(range(1, 11)))
)

print("\nRank integrity:")
print(rank_check)

# Check probability ordering
ordering_check = {}

for target in TARGETS:

    values = top10_by_target[target][
        "Active_Probability"
    ].tolist()

    ordering_check[target] = all(
        values[i] >= values[i + 1]
        for i in range(len(values) - 1)
    )

print("\nProbability descending-order check:")
print(ordering_check)

# Final QC
top10_qc_pass = (
    top10_combined["Target"].nunique() == len(TARGETS)
    and len(top10_combined) == len(TARGETS) * 10
    and top10_combined.isna().sum().sum() == 0
    and rank_check.all()
    and all(ordering_check.values())
)

print("\n" + "=" * 70)

if top10_qc_pass:
    print("✓ TOP-10 PROBABILITY RANKING QC PASSED")
else:
    print("✗ TOP-10 PROBABILITY RANKING QC FAILED")

print("=" * 70)
print("No new predictions performed.")
print("No probability threshold applied.")
print("No source data modified.")
print("=" * 70)
```

```python
# ============================================================
# VS-04 — CELL 30
# Probability-prioritised class-0 candidates
# ============================================================

print("=" * 70)
print("VS-04 — PROBABILITY-PRIORITISED CLASS-0 CANDIDATES")
print("=" * 70)

# ------------------------------------------------------------
# Candidate threshold
# ------------------------------------------------------------

PROBABILITY_CANDIDATE_THRESHOLD = 0.30

print(
    "\nCandidate probability threshold:",
    PROBABILITY_CANDIDATE_THRESHOLD
)

# ------------------------------------------------------------
# Build candidate tables for each target
# ------------------------------------------------------------

near_hit_by_target = {}

for target in TARGETS:

    # Merge probability and predicted class
    target_candidates = probability_matrix[
        [
            "Combined_ID",
            "Combined_Name",
            "Prediction_Source",
            target
        ]
    ].copy()

    target_candidates = target_candidates.rename(
        columns={
            target: "Active_Probability"
        }
    )

    target_candidates["Predicted_Class"] = (
        activity_matrix
        .set_index("Combined_ID")
        .loc[
            target_candidates["Combined_ID"],
            target
        ]
        .values
    )

    # Keep ONLY class-0 compounds at/above threshold
    target_candidates = target_candidates[
        (target_candidates["Predicted_Class"] == 0)
        &
        (
            target_candidates["Active_Probability"]
            >= PROBABILITY_CANDIDATE_THRESHOLD
        )
    ].copy()

    target_candidates = target_candidates.sort_values(
        "Active_Probability",
        ascending=False
    ).reset_index(drop=True)

    target_candidates.insert(
        0,
        "Rank",
        range(1, len(target_candidates) + 1)
    )

    near_hit_by_target[target] = target_candidates

    print("\n" + "-" * 70)
    print(f"TARGET: {target}")
    print("-" * 70)

    print(
        "Number of class-0 candidates:",
        len(target_candidates)
    )

    if len(target_candidates) > 0:
        display(target_candidates)
    else:
        print("None")

# ------------------------------------------------------------
# Combined candidate table
# ------------------------------------------------------------

near_hit_combined = pd.concat(
    [
        table.assign(Target=target)
        for target, table in near_hit_by_target.items()
        if len(table) > 0
    ],
    ignore_index=True
)

if len(near_hit_combined) > 0:

    near_hit_combined = near_hit_combined[
        [
            "Target",
            "Rank",
            "Combined_ID",
            "Combined_Name",
            "Prediction_Source",
            "Predicted_Class",
            "Active_Probability"
        ]
    ]

else:

    near_hit_combined = pd.DataFrame(
        columns=[
            "Target",
            "Rank",
            "Combined_ID",
            "Combined_Name",
            "Prediction_Source",
            "Predicted_Class",
            "Active_Probability"
        ]
    )

# ------------------------------------------------------------
# Summary
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("PROBABILITY-PRIORITISED CLASS-0 SUMMARY")
print("=" * 70)

print(
    "Total candidate associations:",
    len(near_hit_combined)
)

print(
    "Unique candidate compounds:",
    near_hit_combined["Combined_ID"].nunique()
)

if len(near_hit_combined) > 0:

    print("\nCandidates by target:")
    print(
        near_hit_combined["Target"]
        .value_counts()
        .reindex(TARGETS, fill_value=0)
    )

    print("\nHighest candidate probability:")
    display(
        near_hit_combined.loc[
            near_hit_combined["Active_Probability"].idxmax()
        ].to_frame().T
    )

# ------------------------------------------------------------
# Integrity checks
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("CELL 30 QC")
print("=" * 70)

if len(near_hit_combined) > 0:

    print(
        "\nPredicted class values:"
    )
    print(
        near_hit_combined[
            "Predicted_Class"
        ].value_counts()
    )

    print(
        "\nProbabilities below threshold:"
    )
    print(
        (
            near_hit_combined[
                "Active_Probability"
            ]
            < PROBABILITY_CANDIDATE_THRESHOLD
        ).sum()
    )

    print(
        "\nProbabilities outside [0, 1]:"
    )
    print(
        (
            (
                near_hit_combined[
                    "Active_Probability"
                ] < 0
            )
            |
            (
                near_hit_combined[
                    "Active_Probability"
                ] > 1
            )
        ).sum()
    )

    print(
        "\nDuplicate compound-target associations:"
    )
    print(
        near_hit_combined.duplicated(
            subset=[
                "Combined_ID",
                "Target"
            ]
        ).sum()
    )

# ------------------------------------------------------------
# Final QC
# ------------------------------------------------------------

qc_pass = True

if len(near_hit_combined) > 0:

    qc_pass = (
        near_hit_combined[
            "Predicted_Class"
        ].eq(0).all()
        and
        (
            near_hit_combined[
                "Active_Probability"
            ]
            >= PROBABILITY_CANDIDATE_THRESHOLD
        ).all()
        and
        (
            near_hit_combined[
                "Active_Probability"
            ].between(0, 1).all()
        )
        and
        near_hit_combined.duplicated(
            subset=[
                "Combined_ID",
                "Target"
            ]
        ).sum() == 0
    )

print("\n" + "=" * 70)

if qc_pass:
    print("✓ CELL 30 QC PASSED")
else:
    print("✗ CELL 30 QC FAILED")

print("=" * 70)
print("These are NOT predicted hits.")
print("They remain Predicted_Class = 0.")
print("No model predictions performed.")
print("No source data modified.")
print("=" * 70)
```

```python
# ============================================================
# VS-04 — CELL 31
# FINAL COMPOUND-LEVEL SCREENING SUMMARY
# ============================================================

print("=" * 70)
print("VS-04 — FINAL COMPOUND-LEVEL SCREENING SUMMARY")
print("=" * 70)

# ------------------------------------------------------------
# 1. Start from the validated combined activity matrix
# ------------------------------------------------------------

final_summary = activity_matrix[
    [
        "Combined_ID",
        "Combined_Name",
        "Prediction_Source"
    ] + TARGETS
].copy()

# ------------------------------------------------------------
# 2. Number of predicted targets per compound
# ------------------------------------------------------------

final_summary["Number_of_Predicted_Targets"] = (
    final_summary[TARGETS]
    .sum(axis=1)
)

# ------------------------------------------------------------
# 3. Predicted receptor list
# ------------------------------------------------------------

def get_predicted_targets(row):
    return ", ".join(
        [
            target
            for target in TARGETS
            if row[target] == 1
        ]
    )

final_summary["Predicted_Targets"] = (
    final_summary.apply(
        get_predicted_targets,
        axis=1
    )
)

# ------------------------------------------------------------
# 4. Assign primary screening tier
# ------------------------------------------------------------

def assign_screening_tier(row):

    n_targets = row["Number_of_Predicted_Targets"]

    if n_targets >= 1:
        return "Predicted hit"

    return "No predicted activity"

final_summary["Screening_Tier"] = (
    final_summary.apply(
        assign_screening_tier,
        axis=1
    )
)

# ------------------------------------------------------------
# 5. Add highest active probability
# ------------------------------------------------------------

final_summary["Highest_Active_Probability"] = 0.0
final_summary["Highest_Probability_Target"] = ""

for idx, row in final_summary.iterrows():

    compound_id = row["Combined_ID"]

    probability_row = probability_matrix[
        probability_matrix["Combined_ID"] == compound_id
    ]

    if len(probability_row) != 1:
        raise ValueError(
            f"Expected exactly one probability row for {compound_id}, "
            f"found {len(probability_row)}"
        )

    probabilities = probability_row.iloc[0]

    target_probabilities = {
        target: float(probabilities[target])
        for target in TARGETS
    }

    highest_target = max(
        target_probabilities,
        key=target_probabilities.get
    )

    final_summary.loc[
        idx,
        "Highest_Active_Probability"
    ] = target_probabilities[highest_target]

    final_summary.loc[
        idx,
        "Highest_Probability_Target"
    ] = highest_target

# ------------------------------------------------------------
# 6. Reorder columns
# ------------------------------------------------------------

final_summary = final_summary[
    [
        "Combined_ID",
        "Combined_Name",
        "Prediction_Source",
        "Predicted_Targets",
        "Number_of_Predicted_Targets",
        "Highest_Probability_Target",
        "Highest_Active_Probability",
        "Screening_Tier"
    ]
]

# ------------------------------------------------------------
# 7. Display predicted-hit compounds first
# ------------------------------------------------------------

final_summary = final_summary.sort_values(
    by=[
        "Number_of_Predicted_Targets",
        "Highest_Active_Probability"
    ],
    ascending=[
        False,
        False
    ]
).reset_index(drop=True)

print("\nFINAL COMPOUND-LEVEL SUMMARY:")
display(final_summary)

# ------------------------------------------------------------
# 8. Summary statistics
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("FINAL SUMMARY STATISTICS")
print("=" * 70)

print(
    "Total compounds:",
    len(final_summary)
)

print(
    "Compounds with ≥1 predicted target:",
    (
        final_summary["Number_of_Predicted_Targets"] >= 1
    ).sum()
)

print(
    "Compounds with no predicted targets:",
    (
        final_summary["Number_of_Predicted_Targets"] == 0
    ).sum()
)

print("\nScreening tier:")
print(
    final_summary["Screening_Tier"]
    .value_counts()
)

print("\nPrediction source:")
print(
    final_summary["Prediction_Source"]
    .value_counts()
)

print("\n" + "=" * 70)
print("PREDICTED-HIT COMPOUND SUMMARY")
print("=" * 70)

predicted_hit_compounds = final_summary[
    final_summary["Number_of_Predicted_Targets"] >= 1
].copy()

display(predicted_hit_compounds)

# ------------------------------------------------------------
# 9. Integrity QC
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("CELL 31 — FINAL SUMMARY QC")
print("=" * 70)

print("\nRows:", len(final_summary))

print(
    "Duplicate Combined_ID values:",
    final_summary["Combined_ID"].duplicated().sum()
)

print("\nMissing values:")
print(
    final_summary.isna().sum()
)

print("\nInvalid target counts:")
print(
    (
        (final_summary["Number_of_Predicted_Targets"] < 0)
        |
        (final_summary["Number_of_Predicted_Targets"] > len(TARGETS))
    ).sum()
)

print("\nHighest probability range:")
print(
    f"{final_summary['Highest_Active_Probability'].min():.6f}"
    " - "
    f"{final_summary['Highest_Active_Probability'].max():.6f}"
)

# ------------------------------------------------------------
# Expected values
# ------------------------------------------------------------

expected_rows = 290
expected_predicted_hit_compounds = 5

qc_pass = (
    len(final_summary) == expected_rows
    and final_summary["Combined_ID"].duplicated().sum() == 0
    and final_summary.isna().sum().sum() == 0
    and (
        final_summary["Number_of_Predicted_Targets"]
        .between(0, len(TARGETS))
        .all()
    )
    and (
        final_summary[
            "Highest_Active_Probability"
        ]
        .between(0, 1)
        .all()
    )
    and (
        len(predicted_hit_compounds)
        == expected_predicted_hit_compounds
    )
)

print("\n" + "=" * 70)

if qc_pass:
    print("✓ CELL 31 FINAL SUMMARY QC PASSED")
else:
    print("✗ CELL 31 FINAL SUMMARY QC FAILED")

print("=" * 70)
print("No new predictions performed.")
print("No thresholds applied.")
print("No source data modified.")
print("=" * 70)
```
